## 1. Build a minimal MuJoCo scene

A tiny MJCF with a single free body standing in for an object to pick.

```python
import mujoco
import numpy as np

MJCF = """
<mujoco>
  <worldbody>
    <light diffuse=".5 .5 .5" pos="0 0 3" dir="0 0 -1"/>
    <geom type="plane" size="1 1 0.1" rgba=".9 .9 .9 1"/>
    <body name="object" pos="0 0 0.1">
      <joint type="free"/>
      <geom type="box" size="0.03 0.03 0.03" rgba="0.8 0.2 0.2 1"/>
    </body>
  </worldbody>
</mujoco>
"""

model = mujoco.MjModel.from_xml_string(MJCF)
data = mujoco.MjData(model)
print("model loaded, nq =", model.nq)
```

## 2. Roll out a few steps and log with rerun.io

In Colab there's no local rerun viewer, so we save to a `.rrd` file and
download/inspect it afterwards (or open it with the rerun web viewer).

```python
import rerun as rr

rr.init("example01_pick_and_place")

for step in range(200):
    mujoco.mj_step(model, data)
    rr.set_time_sequence("step", step)
    rr.log("object/qpos_z", rr.Scalar(data.qpos[2]))

rr.save("episode_001.rrd")
print("saved episode_001.rrd")
```
